## Q7. Sum of weights (Matrix Operations)
1. Відбираємо всі авто з Азії.
2. Залишаємо колонки `vehicle_weight` та `model_year`.
3. Беремо перші 7 значень і перетворюємо їх у масив NumPy `X`.
4. Знаходимо матрицю `XTX` (множення `X.T` на `X`), а потім її обернену матрицю.

```python
df_Asia = df[df['origin'] == 'Asia']
df_Asia.head()
```

```python
df_2 = df_Asia[['vehicle_weight', 'model_year']]
df_2.head()
```

```python
df_2.iloc[:7]
```

```python
import numpy as np
```

```python
X = np.array(df_2.iloc[:7])
X.shape
```

```python
XTX = X.T.dot(X)
XTX.shape
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv@X.T@y
w
```

```python
w.sum()
```
